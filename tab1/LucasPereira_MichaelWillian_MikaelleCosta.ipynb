### PARTE I: Conceitos de Programação Funcional

#### 1.1. Defina os seguintes conceitos como são compreendidos no paradigma funcional: 
* funções anônimas, funções de primeira ordem, funções de alta ordem, funções recursivas, funções puras, avaliação preguiçosa, _monads_ e estruturas de dados recursivas.

**Funções anônimas** são funções definidas sem um nome associado, geralmente expressas de forma literal no ponto em que são utilizadas. Elas são comuns como argumentos para outras funções ou para uso pontual sem necessidade de reuso.

**Funções de primeira ordem** são funções que operam apenas sobre dados convencionais, ou seja, recebem valores comuns como argumentos e retornam valores comuns. Além disso, não manipulam outras funções.

**Funções de alta ordem** são funções que tratam funções como cidadãs de primeira classe: podem receber uma ou mais funções como argumento, retornar uma função como resultado, ou ambos. Exemplos clássicos são: map, filter e fold/reduce.

**Funções recursivas** são funções que se invocam a si mesmas como parte de sua definição, resolvendo um problema ao reduzi-lo a instâncias menores do mesmo problema até atingir um caso base. No paradigma funcional substituem os laços de repetição.

**Funções puras** são funções que, dados os mesmos argumentos, sempre produzem o mesmo resultado e não causam efeitos colaterais observáveis, ou seja, não alteram estado externo, não realizam I/O e não dependem de variáveis globais mutáveis.

**Avaliação preguiçosa** (lazy evaluation) é uma estratégia de avaliação em que expressões só são computadas no momento em que seus valores são efetivamente necessários. Ela permite trabalhar com estruturas potencialmente infinitas e evitar cálculos desnecessários.

**Monads** são uma estrutura algébrica que encapsula um valor em um contexto computacional (como possibilidade de falha, estado, I/O ou não-determinismo) e fornece operações padronizadas (tipicamente unit/return e bind/>>= ) para encadear computações dentro desse contexto sem violar a pureza funcional.

**Estruturas de dados recursivas** são estruturas definidas em termos de si mesmas, onde cada instância pode conter referências a instâncias do mesmo tipo. Listas encadeadas e árvores são os exemplos canônicos; a recursividade da estrutura corresponde naturalmente à recursividade das funções que a processam.

#### 2.1. Rolar dados

#### 2.2. Contar ocorrências
Crie a função ```contar(lst, alvo)``` que retorna o número de vezes que _alvo_ ocorre em _lst_, usando recursão.
Exemplo:
```python
contar([1, 5, 1, 1, 5, 6], 1) -> 3
contar([1, 5, 1, 1, 5, 2], 2) -> 1
```

````kotlin
contar = lambda lst, alvo : 0 if not lst else ( 1 if lst[0] == alvo else 0) + contar(lst[1:], alvo)
````

#### 2.4. Pontos por número e dados pontuadores
Crie a função ```pontos_por_numero_(dados, num)``` que considerando um arranjo dos seis dados, retorna a pontuação referente a _num_ e quantos foram os dados pontuadores. Considere que o arranjo nunca pode ser formado pelos números de 1 a 6, cada um deles ocorrendo uma única vez, como em [1, 2, 3, 4, 5, 6]. 
Exemplo:
```python
pontos_por_numero([1, 2, 3, 4, 5, 6], 1) -> (100, 1) # O numero 1 ocorre 1x no arranjo. Portanto, para 1, a pontuação do arranjo é 100 com 1 dado pontuando.
pontos_por_numero([1, 2, 3, 4, 5, 6], 5) -> (50, 1) # O numero 5 ocorre 1x no arranjo. Portanto, para 5, a pontuação do arranjo é 50 com 1 dado pontuando.
pontos_por_numero([1, 5, 1, 1, 5, 6], 1) -> (1000, 3) # O numero 1, ocorre 3x no arranjo. Portanto, para 1, a pontuação do arranjo é 1000 com 3 dados pontuando.
pontos_por_numero([1, 5, 1, 1, 5, 6], 5) -> (100, 2) # O numero 5, ocorre 2x no arranjo. Portanto, para 5, a pontuação do arranjo é 100 com 2 dados pontuando.
pontos_por_numero([1, 5, 1, 1, 5, 6], 6) -> (0, 0) # O numero  6, ocorre 1x no arranjo. Portanto, para 6, a pontuação do arranjo é 0 com nenhum dado pontuando.
pontos_por_numero([1, 1, 1, 1, 1, 1], 1) -> (2000, 6) # O numero 1, ocorre 6x no arranjo. Portanto, para 1, a pontuação do arranjo é 2000 com 6 dados pontuando.
pontos_por_numero([5, 5, 5, 5, 5, 5], 5) -> (1000, 6) # O numero 5, ocorre 6x no arranjo. Portanto, para 5, a pontuação do arranjo é 1000 com 6 dados pontuando.
pontos_por_numero([2, 2, 2, 2, 2, 2], 2) -> (400, 6) # O numero 2, ocorre 6x no arranjo. Portanto, para 2, a pontuação do arranjo é 400 com 6 dados pontuando.
```

````kotlin
pontos_por_numero = lambda dados, num: (
    ((contar(dados, num) // 3) * 1000 + (contar(dados, num) % 3) * 100, contar(dados, num)) if num == 1 else
    ((contar(dados, num) // 3) * 500 + (contar(dados, num) % 3) * 50, contar(dados, num )) if num == 5 else
    ((contar(dados, num) // 3) * (100 * num), (contar(dados, num) // 3) * 3)
)
````

#### 2.5. Pontuação da jogada
Crie a função ```pontuacao_jogada_(dados)``` que considerando um arranjo dos seis dados, retorna a pontuação referente ao arranjo bem como o número de dados pontuadores. Defina ```pontuacao_jogada``` em função de ```pontos_por_numero```. 
Exemplo:
```python
pontuacao_jogada([1, 5, 2, 3, 4, 6]) -> (1500, 6) # todos os números de 1 a 6, pontuação de 1500 com todos os dados (6) usados em pontuação
pontuacao_jogada([2, 2, 3, 4, 4, 6]) -> (0, 0) # 0 pontos, nenhum dado usado em pontuação
pontuacao_jogada([1, 1, 1, 2, 3, 4]) -> (1000, 3) # 3 dados pontuaram (a trinca de 1 que resulta em 1000 pontos) 
pontuacao_jogada([2, 2, 2, 4, 5, 6]) -> (250, 4) # 4 dados pontuaram (a trinca de 2 e o numero 5, resultando em 250 pontos) 
pontuacao_jogada([1, 5, 2, 2, 3, 4]) -> (150, 2) # 2 dados pontuaram (1 e 5, resultando em 150 pontos) 
pontuacao_jogada([2, 2, 2, 2, 2, 2]) -> (400, 6) # 6 dados pontuaram (2 trincas de 2, resultando em 400 pontos)
pontuacao_jogada([2, 2, 2, 3, 3, 3]) -> (500, 6) # 6 dados pontuaram (trinca de 2 e trinca de 3, resultando em 500 pontos)
pontuacao_jogada([4, 4, 4, 6, 6, 6]) -> (1000, 6) # 6 dados pontuaram (trinca de 4 e trinca de 6, resultando em 1000 pontos)
pontuacao_jogada([5, 6, 5, 6, 6, 6]) -> (700, 5) # 5 dados pontuaram (dois 5 e trinca de 6, resultando em 700 pontos) -- note que um dos 6 não pontuou
```

````kotlin
pontuacao_jogada = lambda dados: (1500, 6) if sorted(dados) == [1, 2, 3, 4, 5, 6] else (
    sum(pontos_por_numero(dados, n)[0] for n in range(1, 7)),
    sum(pontos_por_numero(dados, n)[1] for n in range(1, 7))
)
````

#### 2.6. Turno
Crie a função _recursiva_ ```turno(pontos_rodada, n_dados, escolha, verbose = True)``` que implementa um turno de jogadas de um jogador. Use as funções ```rolar_dados``` para rolar os dados; ```pontuacao_jogada``` para obter pontos da jogada e número de dados pontuadores; e a função ```escolha``` para definir a escolha do usuário, como descrito melhor abaixo. A função tem como parâmetros:
* ```pontos_rodada```: número de pontos acumulados pelo jogador até a rodada atual
* ```n_dados```: número de dados a serem jogados
* ```escolha```: função usada para decidir se o jogador para ou continua jogando na rodada. A decisão é por parar se ```escolha``` retorna ```False```. É continuar se ```escolha``` retorna ```True```. Esta função deve tomar a decisão com base no número de dados restantes e na pontuação atual do jogador. Abaixo, um exemplo de escolha aleatória (que, por ser aleatória, ignora os parâmetros :)):
```python
def escolha_aleatoria(n_restante, pontos):
    return random.choice([True, False])
```
* ```verbose = False```: parâmetro opcional que indica se mensagens de status serão exibidas ou não, indicando o que ocorre no turno.

````kotlin
def escolha_aleatoria(n_restante, pontos):
    return random.choice([True, False])

def turno(pontos_rodada, n_dados, escolha, verbose=True):
    # Rolar os dados atuais
    dados = rolar_dados(n_dados)
    
    # Calcular pontuação e dados usados no lançamento
    pts_lancamento, usados = pontuacao_jogada(dados)
    total_acumulado = pontos_rodada + pts_lancamento

    # Calcular dados restantes e aplicar regra de hot dice 
    # Se todos os dados foram usados, o jogador pode rolar 6 novamente 
    dados_restantes = 6 if n_dados - usados == 0 else n_dados - usados
    
    if verbose:
        print(f"Lança {dados} ({pts_lancamento} pts) | Acumulado: {total_acumulado} | Restantes: {dados_restantes} dados")

    # Condição de farfle (zero pontos no lançamento)
    if pts_lancamento == 0:
        if verbose:
            print("FARKLE!!!! Você perdeu todos os pontos acumulados na rodada.")
        return 0 # O jogador perde tudo o que acumulou nesta rodada [cite: 31]
    
    # Decisão do jogador (parar ou continuar)
    # A função 'escolha' retorna False para parar e True para continuar
    if not escolha(dados_restantes, total_acumulado):
        if verbose:
            print(f"Jogador PARA com {total_acumulado} pontos.")
        return total_acumulado

    # Chamada recursiva 
    if verbose:
        print(f"Continua com {dados_restantes} dados restantes.")
    return turno(total_acumulado, dados_restantes, escolha, verbose)
````

#### 2.7. Escolha heurística
Crie a função ```escolha_heuristica(n_restante, pontos)``` que, ao contrário da função ```escolha_aleatória```, efetivamente usa a informação de número restante de dados e pontos atuais do jogador para escolhar se para ou continua. Projete uma heurística que tenha uma probabilidade maior de fazer mais pontos que a função de escolha aleatória ao longo do tempo.

````kotlin
def escolha_heuristica(n_restante, pontos):
    # Regra de Ouro: Se temos 6 dados (Hot Dice), sempre continuamos.
    # A chance de pontuar com 6 dados é de +/- 97.7%.
    if n_restante >= 5:
        return True
    
    # Com 4 dados, a chance de sucesso é +/- 77%. Arriscamos até acumular um valor alto.
    if n_restante == 4:
        return pontos < 1000
    
    # Com 3 dados, a chance cai para +/- 55%. Paramos se já tivermos garantido uma trinca.
    if n_restante == 3:
        return pontos < 600
    
    # Com 2 dados, a chance é de +/- 44%. Só arriscamos se tivermos quase nada.
    if n_restante == 2:
        return pontos < 300
    
    # Com 1 dado, a chance é de apenas 33.3% (apenas se cair 1 ou 5).
    # O risco de perder tudo (Farkle) é de 66.6%.
    # Só arriscamos se os pontos na rodada forem irrelevantes (ex: < 150).
    return pontos < 150
````

#### 2.8. Simulando 1000 Partidas
A função _recursiva_ ```partida(estrategia0, estrategia1, pontos = [0, 0], jogador = 0, pontuacao_max = 10000)``` simula uma partida entre dois jogadores.

```python
def partida(estrategia0, estrategia1, pontos=[0, 0], jogador=0, pontuacao_max=10000, verbose=False):
    '''
        * estrategia0: função de escolha para jogador 0
        * estrategia1: função de escolha para jogador 1
        * pontos = [0, 0]: pontuação do jogador 0 e do jogador 1 
          (0 e 0 no início da partida)
        * jogador = 0: jogador que vai iniciar a partida (0, por default)
        * pontuacao_max = 10000: pontuação máxima que determina fim de jogo 
          (10000 por default)
    '''
    # Condição de término
    if pontos[0] >= pontuacao_max or pontos[1] >= pontuacao_max:
        if verbose:
            print(f"\nFIM DE JOGO! Placar final: Jogador 0: {pontos[0]} | Jogador 1: {pontos[1]}")
        if pontos[0] >= pontuacao_max and pontos[1] >= pontuacao_max:
            if pontos[0] > pontos[1]:
                return 0
            elif pontos[1] > pontos[0]:
                return 1
            else:
                return None
        elif pontos[0] >= pontuacao_max:
            return 0
        else:
            return 1

    # Turno do jogador atual
    if verbose:
        print(f"\n- Jogador {jogador} (Placar: {pontos[0]}-{pontos[1]}) ---")
    
    escolha = estrategia0 if jogador == 0 else estrategia1
    pts_rodada = turno(0, 6, escolha, verbose)    
    novos_pontos = [pontos[0] + (pts_rodada if jogador == 0 else 0),
                    pontos[1] + (pts_rodada if jogador == 1 else 0)]
    
    return partida(estrategia0, estrategia1, novos_pontos, 
                   1 - jogador, pontuacao_max, verbose)
```

Use o código abaixo para simular 1000 partidas entre sua estratégia heurística e uma estratégia aleatória. Qual o percentual de vitórias da sua estratégia?

```python
def simular(n=1000):
```

````kotlin
def partida(estrategia0, estrategia1, pontos=[0, 0], jogador=0, pontuacao_max=10000, verbose=True):

    # Condição de término (alguém atingiu a pontuação máxima) 
    if pontos[0] >= pontuacao_max or pontos[1] >= pontuacao_max:
        if verbose:
            print(f"\nFIM DE JOGO! Placar final: Jogador 0: {pontos[0]} Jogador 1: {pontos[1]}")
        
        # Retorna o vencedor
        if pontos[0] > pontos[1]: return 0
        elif pontos[1] > pontos[0]: return 1
        return None

    # Início do turno atual
    if verbose:
        print(f"\n--- Vez do Jogador {jogador} (Placar: {pontos[0]}-{pontos[1]}) ---")

    # Define qual estratégia usar e executa o turno
    estrategia_atual = estrategia0 if jogador == 0 else estrategia1
    pts_rodada = turno(0, 6, estrategia_atual, verbose)

    # Atualiza os pontos criando uma nova lista (Imutabilidade) [cite: 194-195]
    novos_pontos = [
        pontos[0] + (pts_rodada if jogador == 0 else 0),
        pontos[1] + (pts_rodada if jogador == 1 else 0)
    ]

    # Chamada recursiva: alterna o jogador (1 - jogador) para a próxima rodada
    return partida(estrategia0, estrategia1, novos_pontos, 1 - jogador, pontuacao_max, verbose)

def simular(n=1000):
    vitorias = 0
    # O enunciado permite o uso de range/for nesta função específica
    for _ in range(n):
        # A nossa estratégia heurística é o Jogador 0
        vencedor = partida(escolha_heuristica, escolha_aleatoria, verbose=False)
        if vencedor == 0:
            vitorias += 1
            
    percentual = (vitorias / n) * 100
    return f"Heurística venceu {percentual:.2f}% das partidas contra aleatório."
````

````kotlin
fun passwordClassification(password: String): Pair<String, List<String>>{

    val criteriosNaoAtendidos = listOfNotNull(
        "tamanho".takeIf {password.length < 8},
        "maiúscula".takeIf { password.none { it.isUpperCase()}},
        "minúscula".takeIf{ password.none { it.isLowerCase()}},
        "dígito".takeIf {password.none {it.isDigit()}},
        "caractere especial".takeIf {password.none {it in "!@#$%^&*()-+"}}
    )

    val pontuacao = 5 - criteriosNaoAtendidos.size


    val classificacao = when(pontuacao){
        0, 1 -> "Fraca"
        2, 3 -> "Moderada" 
        4 -> "Forte"
        5 -> "Muito Forte"
        else -> "Invalida"
    }

    return Pair(classificacao, criteriosNaoAtendidos)
}


fun main(){
    val senhas = listOf(
        "abc",
        "Abc123",
        "Abc123!@",
        "ABCDEFGH",
        "Abcdefg1",
        "12345678"
    )

    senhas.forEach{ senha ->
        println("$senha -> ${passwordClassification(senha)}")
    }
}

main()
````

#### 3.2. Tradução para Kotlin com LLM
Utilize um LLM (ex.: ChatGPT, Gemini, Deepseek) para traduzir seu código Python final da parte II (_Farkle_) para Kotlin, mantendo o estilo funcional.

O LLM utilizado foi: Gemini

````kotlin
import kotlin.random.Random

// Rolar dados
val rolarDados: (Int) -> List<Int> =
    { n -> List(n) { Random.nextInt(1, 7) } }

// Contar ocorrências recursivamente
fun contar(lista: List<Int>, alvo: Int): Int =
    if (lista.isEmpty()) {
        0
    } else {
        (if (lista.first() == alvo) 1 else 0) +
        contar(lista.drop(1), alvo)
    }

// Valores únicos recursivamente
fun valoresUnicos(lista: List<Int>): List<Int> =
    if (lista.isEmpty()) {
        emptyList()
    } else {
        listOf(lista.first()) +
        valoresUnicos(
            lista.drop(1).filter { it != lista.first() }
        )
    }

// Pontuação por número
fun pontosPorNumero(
    dados: List<Int>,
    num: Int
): Pair<Int, Int> {

    val qtd = contar(dados, num)

    return when (num) {

        1 -> Pair(
            (qtd / 3) * 1000 + (qtd % 3) * 100,
            qtd
        )

        5 -> Pair(
            (qtd / 3) * 500 + (qtd % 3) * 50,
            qtd
        )

        else -> Pair(
            (qtd / 3) * (100 * num),
            (qtd / 3) * 3
        )
    }
}

// Pontuação da jogada
fun pontuacaoJogada(
    dados: List<Int>
): Pair<Int, Int> {

    return if (dados.sorted() == listOf(1, 2, 3, 4, 5, 6)) {

        Pair(1500, 6)

    } else {

        Pair(

            (1..6).sumOf {
                pontosPorNumero(dados, it).first
            },

            (1..6).sumOf {
                pontosPorNumero(dados, it).second
            }
        )
    }
}

// Estratégia heurística
fun escolhaHeuristica(
    nRestante: Int,
    pontos: Int
): Boolean {

    return when {

        nRestante >= 5 ->
            true

        nRestante == 4 ->
            pontos < 1000

        nRestante == 3 ->
            pontos < 600

        nRestante == 2 ->
            pontos < 300

        else ->
            pontos < 150
    }
}

// Estratégia aleatória
fun escolhaAleatoria(
    nRestante: Int,
    pontos: Int
): Boolean =
    Random.nextBoolean()

// Turno recursivo
fun turno(
    pontosRodada: Int,
    nDados: Int,
    escolha: (Int, Int) -> Boolean,
    verbose: Boolean = true
): Int {

    // Rolar dados
    val dados = rolarDados(nDados)

    // Pontuação do lançamento
    val (ptsLancamento, usados) =
        pontuacaoJogada(dados)

    val totalAcumulado =
        pontosRodada + ptsLancamento

    // Hot Dice
    val dadosRestantes =
        if (nDados - usados == 0) {
            6
        } else {
            nDados - usados
        }

    if (verbose) {

        println(
            "Lança $dados " +
            "($ptsLancamento pts) | " +
            "Acumulado: $totalAcumulado | " +
            "Restantes: $dadosRestantes dados"
        )
    }

    // Farkle
    if (ptsLancamento == 0) {

        if (verbose) {
            println(
                "FARKLE!!!! " +
                "Você perdeu todos os pontos."
            )
        }

        return 0
    }

    // Jogador decide parar
    if (!escolha(dadosRestantes, totalAcumulado)) {

        if (verbose) {
            println(
                "Jogador PARA com " +
                "$totalAcumulado pontos."
            )
        }

        return totalAcumulado
    }

    // Continua recursivamente
    if (verbose) {
        println(
            "Continua com " +
            "$dadosRestantes dados restantes."
        )
    }

    return turno(
        totalAcumulado,
        dadosRestantes,
        escolha,
        verbose
    )
}

// Partida completa
fun partida(
    estrategia0: (Int, Int) -> Boolean,
    estrategia1: (Int, Int) -> Boolean,
    pontos: List<Int> = listOf(0, 0),
    jogador: Int = 0,
    pontuacaoMax: Int = 10000,
    verbose: Boolean = false
): Int? {

    // Fim de jogo
    if (
        pontos[0] >= pontuacaoMax ||
        pontos[1] >= pontuacaoMax
    ) {

        if (verbose) {

            println(
                "\nFIM DE JOGO! " +
                "Jogador 0: ${pontos[0]} " +
                "Jogador 1: ${pontos[1]}"
            )
        }

        return when {

            pontos[0] > pontos[1] -> 0

            pontos[1] > pontos[0] -> 1

            else -> null
        }
    }

    // Turno atual
    if (verbose) {

        println(
            "\n--- Vez do Jogador " +
            "$jogador " +
            "(Placar: ${pontos[0]}-${pontos[1]}) ---"
        )
    }

    val estrategiaAtual =
        if (jogador == 0)
            estrategia0
        else
            estrategia1

    val ptsRodada =
        turno(0, 6, estrategiaAtual, verbose)

    // Imutabilidade
    val novosPontos = listOf(

        pontos[0] +
        if (jogador == 0) ptsRodada else 0,

        pontos[1] +
        if (jogador == 1) ptsRodada else 0
    )

    // Próxima rodada
    return partida(
        estrategia0,
        estrategia1,
        novosPontos,
        1 - jogador,
        pontuacaoMax,
        verbose
    )
}

// Simulação
fun simular(n: Int = 1000): String {

    var vitorias = 0

    repeat(n) {

        val vencedor = partida(
            ::escolhaHeuristica,
            ::escolhaAleatoria,
            verbose = false
        )

        if (vencedor == 0) {
            vitorias++
        }
    }

    val percentual =
        (vitorias.toDouble() / n) * 100

    return "Heurística venceu " +
           "${"%.2f".format(percentual)}% " +
           "das partidas contra aleatório."
}

// Main
fun main() {

    println(simular())

    println("\n=== PARTIDA EXEMPLO ===")

    partida(
        ::escolhaHeuristica,
        ::escolhaAleatoria,
        verbose = false
    )
}
````

### Simplicidade

No contexto da programação funcional, Python e Kotlin apresentam abordagens diferentes em relação à simplicidade. Python possui uma sintaxe extremamente enxuta e próxima da linguagem natural, o que facilita a escrita de funções puras, uso de map, filter, lambda e compreensões de listas. Isso torna o aprendizado inicial da programação funcional mais simples e intuitivo. Kotlin, por outro lado, exige uma estrutura um pouco mais formal devido à tipagem estática e à necessidade de declarar tipos em algumas situações, mas oferece recursos funcionais muito integrados à linguagem, como funções de ordem superior, funções inline e imutabilidade com val. Assim, Python tende a ser mais simples para prototipação rápida, enquanto Kotlin busca simplicidade aliada a maior segurança estrutural.

### Ortogonalidade

Kotlin apresenta maior ortogonalidade na programação funcional porque seus recursos funcionais são integrados de maneira consistente com o restante da linguagem. Funções podem ser tratadas como valores, passadas como argumentos e combinadas facilmente com classes, coleções e orientação a objetos. Em Python também existe suporte funcional, porém alguns recursos parecem adicionados sobre um núcleo originalmente imperativo e orientado a objetos. Por exemplo, funções lambda em Python possuem limitações em comparação às funções anônimas de Kotlin. Apesar disso, ambas permitem combinar múltiplos paradigmas, mostrando semelhança na flexibilidade, mas Kotlin apresenta uma integração mais uniforme entre paradigmas.

### Tipos de Dados

Uma diferença importante entre Kotlin e Python na programação funcional está no sistema de tipos. Kotlin possui tipagem estática forte, o que aumenta a segurança ao manipular funções e coleções imutáveis, além de reduzir erros em operações funcionais complexas. Python utiliza tipagem dinâmica, permitindo maior flexibilidade e rapidez na escrita do código funcional, mas aumentando a possibilidade de erros aparecerem apenas em tempo de execução. As duas linguagens suportam estruturas imutáveis e funções como cidadãos de primeira classe, porém Kotlin enfatiza mais a segurança e previsibilidade, enquanto Python prioriza liberdade e agilidade.


### Projeto de Sintaxe

Python possui uma sintaxe minimalista que favorece bastante a escrita funcional, especialmente em pipelines simples de transformação de dados. Expressões com listas, geradores e funções lambda tornam o código compacto e legível. Kotlin também oferece uma sintaxe moderna para programação funcional, com destaque para trailing lambdas, extension functions e operadores seguros. Entretanto, sua sintaxe pode parecer mais verbosa devido à tipagem explícita e às estruturas da JVM. As duas linguagens compartilham o objetivo de tornar o código funcional expressivo, mas Python enfatiza concisão extrema, enquanto Kotlin tenta equilibrar clareza, segurança e funcionalidade.

### Suporte à Abstração

Kotlin oferece um suporte muito forte à abstração funcional. A linguagem possui funções de ordem superior, funções inline, extensões e recursos avançados como corrotinas, permitindo criar abstrações poderosas e reutilizáveis. Python também suporta abstrações funcionais através de decorators, closures e funções como objetos de primeira classe. Contudo, Kotlin possui mecanismos mais robustos para garantir consistência nessas abstrações devido à tipagem estática. Em contrapartida, Python oferece mais liberdade e menos restrições, facilitando experimentação rápida, mas podendo gerar soluções menos seguras em projetos grandes.


### Expressividade

As duas linguagens são bastante expressivas na programação funcional, mas de maneiras diferentes. Python se destaca pela simplicidade sintática e pela facilidade de escrever operações funcionais curtas e legíveis. Já Kotlin oferece grande expressividade através de recursos modernos, como extension functions e chaining de operações em coleções. Em Kotlin, operações funcionais costumam ficar muito elegantes e seguras ao mesmo tempo. Em Python, a expressividade vem mais da naturalidade da sintaxe. Assim, ambas permitem escrever código funcional compacto, mas Kotlin combina expressividade com controle de tipos, enquanto Python favorece fluidez e rapidez.

### Checagem de Tipos

Na programação funcional, a checagem de tipos é um dos maiores diferenciais entre Kotlin e Python. Kotlin realiza verificação de tipos em tempo de compilação, reduzindo muitos erros antes da execução do programa. Isso é especialmente útil em cadeias de funções e operações sobre coleções. Python realiza a verificação apenas em tempo de execução, tornando o desenvolvimento mais flexível, mas menos seguro. Embora Python tenha introduzido type hints, eles não possuem a mesma rigidez do sistema de tipos de Kotlin. Dessa forma, Kotlin oferece maior confiabilidade em aplicações funcionais complexas, enquanto Python oferece mais liberdade ao programador.

### Manipulação de Exceções

Kotlin possui uma abordagem mais estruturada para manipulação de exceções, permitindo integrar tratamento de erros com recursos funcionais como Result e operadores seguros. Isso facilita escrever código funcional mais previsível e menos sujeito a falhas inesperadas. Python utiliza tratamento de exceções de maneira mais flexível e dinâmica, com blocos try/except, mas sem mecanismos funcionais tão integrados quanto Kotlin. As duas linguagens permitem lidar com erros em funções, porém Kotlin favorece soluções mais controladas e seguras, enquanto Python prioriza simplicidade e praticidade.

### Restrição de Aliases

Na programação funcional, a imutabilidade é importante para evitar efeitos colaterais causados por aliases. Kotlin incentiva fortemente a imutabilidade através do uso de val e coleções imutáveis, reduzindo problemas relacionados ao compartilhamento de referências. Python permite trabalhar com estruturas imutáveis, como tuplas, mas listas e dicionários mutáveis são muito comuns, aumentando o risco de alterações inesperadas em funções. Assim, Kotlin possui um controle maior sobre aliases e efeitos colaterais, enquanto Python depende mais da disciplina do programador para manter princípios funcionais.


### Legibilidade

Python é amplamente reconhecido por sua alta legibilidade, característica que beneficia bastante a programação funcional em códigos pequenos e médios. Sua sintaxe limpa torna operações funcionais fáceis de entender. Kotlin também apresenta boa legibilidade, especialmente em pipelines funcionais com coleções e lambdas, mas sua sintaxe pode parecer mais densa para iniciantes devido à tipagem e recursos adicionais. Ambas as linguagens permitem escrever código funcional claro, porém Python tende a ser mais acessível visualmente, enquanto Kotlin prioriza clareza associada à segurança.

### Escrita

Em termos de escrita de código funcional, Python normalmente exige menos código para implementar soluções simples, tornando o desenvolvimento mais rápido. Kotlin, embora um pouco mais verboso, oferece ferramentas modernas que reduzem repetição e facilitam manutenção, como inferência de tipos e funções de extensão. Em Python, escrever código funcional é muito direto, enquanto Kotlin exige mais atenção à estrutura do programa. Ainda assim, Kotlin oferece vantagens em projetos maiores por manter organização e consistência mais facilmente.

### Confiabilidade

Kotlin apresenta maior confiabilidade na programação funcional devido à tipagem estática, controle de nulidade e incentivo à imutabilidade. Esses recursos ajudam a evitar erros comuns em aplicações complexas. Python, apesar de extremamente produtivo, depende mais de testes e da experiência do programador para garantir robustez, já que muitos erros só aparecem durante a execução. Ambas as linguagens suportam conceitos funcionais importantes, mas Kotlin é mais adequado para sistemas que exigem alta segurança e manutenção de longo prazo, enquanto Python se destaca em prototipação rápida e desenvolvimento ágil.
