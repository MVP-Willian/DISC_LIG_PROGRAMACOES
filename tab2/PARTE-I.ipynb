```kotlin
class Matriz2D(dados: List<List<Int>>) {

    //atributos
    private val dados: List<List<Int>> = dados.map { it.toList() }
    val linhas: Int = this.dados.size
    val colunas: Int = this.dados.first().size

    //Verificações ao instanciar um objeto do tipo Matriz2D
    init {
        require(this.dados.isNotEmpty()) { "Matriz não pode ser vazia" }
        val primeiraLinha = this.dados.first().size
        require(this.dados.all { it.size == primeiraLinha }) {
             "Todas as linhas devem ter o mesmo tamanho"
        }
    }
    // Operadores
    operator fun plus(outra: Matriz2D): Matriz2D {
        require(
            this.linhas == outra.linhas && this.colunas == outra.colunas 
        ) { "As matrizes precisam ter dimensões iguais"}
        val matrizResultado = this.dados.zip(outra.dados) { linhaA, linhaB -> linhaA.zip(linhaB) { a, b -> a + b}}
        return Matriz2D(matrizResultado)
    } 


    operator fun minus(outra: Matriz2D): Matriz2D {
        require(
            this.linhas == outra.linhas && this.colunas == outra.colunas
        ) { "As matrizes precisam ter dimensões iguais"}
        val matrizResultado = this.dados.zip(outra.dados) { linhaA, linhaB -> linhaA.zip(linhaB) { a, b -> a - b}}
        return Matriz2D(matrizResultado)
    }


    operator fun times(escalar: Int): Matriz2D {
        val matrizResultado = this.dados.map { linha -> linha.map { elemento -> elemento * escalar}}
        return Matriz2D(matrizResultado)
    }


    operator fun times(outra: Matriz2D): Matriz2D {
        require(
            this.colunas == outra.linhas
        ) { "O tamanho de colunas da Primeira matriz, precisa ser o mesma do tamanho de linhas da segunda matriz"}
   
        val matrizResultado = List(this.linhas) { i ->
            List(outra.colunas) { 
                j -> List(this.colunas){ 
                    k -> this.dados[i][k] * outra.dados[k][j]
                    }. sum()
                }
            }
        return Matriz2D(matrizResultado)
    }



    //metodos
    fun transpor(): Matriz2D {
        var dadosNovos: List<List<Int>> = List(this.linhas) {
           j->  List(this.colunas){
            i ->  this.dados[i][j]
           }
        }

        return Matriz2D(dadosNovos)
    }


    operator fun get(i: Int, j: Int): Int {
        return this.dados[i][j]
    }



    override fun equals(other: Any?): Boolean {
        if (this === other) return true      
        if (other !is Matriz2D) return false    
        return this.dados == other.dados      
    }


    override fun toString(): String = dados.toString()

    fun obterDados(): List<List<Int>> = dados
}

operator fun Int.times(matriz: Matriz2D): Matriz2D = matriz * this

fun main() {
    val a = Matriz2D(listOf(listOf(1,2), listOf(3,4)))
    val b = Matriz2D(listOf(listOf(0, 1), listOf(1, 0)))


    println(a + b)
    println(a - b)
    println(a * b)
    println(2 * a)
    println(a * 3)
    println(a.transpor())
    println(a[0, 1])

    //Exceções:
    //Matriz2D(listOf(listOf(1, 2), listOf(3))) //ok
    //a * Matriz2D(listOf(listOf(1, 2, 3))) //ok
    //Matriz2D(listOf(listOf(1))).det() // ok 
    
}

main()
```

Com base no que eu aprendi com esse ultimo exercicio, sobre kotlin OO e na minha experiencia com python OO, considerando os seguintes critérios:

legibilidade, escrita, confiabiliade
simplicidade, ortogonalidade, tipos de dados, Projeto de sintaxe, suporte à abstracao, Expressividade, checagem de tipos, manipulações de exceções e restrições de aliases

Abaixo tem um paragrafo para cada criterio, destacando semelhancas e diferenças entre kotlin e python no contexto da programação OO.

__Legibilidade__. Na minha experiência, percebo que tanto Kotlin quanto Python são legíveis, mas por caminhos opostos. Em Python eu gosto da ausência de ruído visual: indentação no lugar de chaves, sem declaração de tipo. Já em Kotlin convivo com mais símbolos (chaves, <>, anotações), mas em troca o tipo explícito me comunica a intenção, quando leio private val dados: List<List<Int>> na minha matriz, já sei exatamente o que é aquilo sem rodar nada. Notei que, em OO, Kotlin deixa explícito o que em Python eu deixava implícito na convenção (visibilidade, mutabilidade via val/var). A sobrecarga de operadores me ajudou nos dois lados: a + b ficou igualmente legível. No fim, acho que Python ganha em limpeza visual e Kotlin em intenção explícita.


__Escrita (writability)__. Sinto que escrevo mais rápido em Python: por ser dinâmico e não exigir tipos, eu prototipo na hora. Kotlin me pareceu mais verboso, mas reparei que ele compensa com inferência de tipos, construtor primário e recursos que cortam boilerplate, no meu exercício, se eu tivesse usado uma data class, teria ganhado equals, hashCode e toString de graça, em vez de escrever tudo na mão. Vi que Python tem o mesmo atalho com @dataclass. Para sobrecarregar operadores, usei operator fun plus em Kotlin e sei que em Python seria o dunder __add__, esforço parecido. Concluo que em Python escrevo mais rápido no improviso, e em Kotlin escrevo com mais segurança por linha.

__Confiabilidade__. Aqui eu senti na pele que Kotlin é mais confiável. A tipagem estática pegou erros antes de rodar, lembro do meu .obterDados(), que quebrava o tipo de retorno e o compilador me barrou na hora; em Python esse mesmo deslize só apareceria rodando, ou nem apareceria se eu não exercitasse aquele caminho. Quando usei val e a cópia defensiva (map { it.toList() }), garanti que ninguém corrompesse o estado interno da minha matriz por fora. Entendi que Python é mais flexível, mas eu pagaria essa flexibilidade com erros de tipo e de aliasing aparecendo só em tempo de execução.


__Simplicidade__. Acho Python mais simples de cabeça, tem menos construções, um modelo de objeto bem uniforme ("tudo é objeto") e aquela ideia de "uma forma óbvia de fazer". Kotlin me ofereceu mais peças (val/var, modificadores de visibilidade, companion object, data/sealed class, extension functions, === vs ==), o que enriquece mas me deu mais coisa para aprender. Isso ficou claro no exercício: precisei entender a diferença entre === e ==, e entre List e MutableList, distinções que em Python eu não preciso enfrentar tão cedo. Percebo que Kotlin troca simplicidade por segurança e poder.

__Ortogonalidade__. Sobre o quanto os recursos se combinam sem casos especiais, achei Python bem ortogonal: operadores mapeiam de forma uniforme para os dunders, e funções e classes são objetos de primeira classe. Kotlin também me pareceu razoavelmente ortogonal, as extension functions, por exemplo, me deixaram estender um tipo que eu nem controlo, foi assim que criei o Int.times(Matriz2D), mas percebi que ele carrega exceções vindas do Java: primitivos vs. tipos "boxed", e o === se comportando diferente para inteiros pequenos por causa do cache. Então, para mim, Python é um pouco mais uniforme e Kotlin tem alguns cantos com regras próprias.


__Tipos de dados__. Esse critério foi central no meu exercício. Vi que Kotlin tem um sistema de tipos rico e estático: genéricos de verdade, tipos anuláveis (Int?) e, o que mais me marcou, a separação entre coleções somente-leitura (List) e mutáveis (MutableList). Foi justamente essa distinção que me obrigou a fazer a cópia defensiva. Já Python é dinâmico: tudo é objeto, os genéricos só existem como type hints não checados, e a granularidade de mutabilidade é mais grossa (list vs. tuple). Concluo que Kotlin me deu mais ferramentas de tipo para expressar minha intenção sobre mutabilidade.

__Projeto de sintaxe__. Notei duas filosofias diferentes e coerentes. Kotlin é da família C (chaves, ponto e vírgula opcional), mas modernizado: corpo de expressão (fun x() = ...), lambda no final da chamada e construtor primário enxuto (class Matriz2D(dados: ...)). Python usa indentação significativa e blocos com dois-pontos, o que acho muito limpo. Um ponto que gostei na sintaxe de Kotlin foi o operator fun plus, autoexplicativo, enquanto o __add__ de Python depende de eu conhecer os nomes mágicos. Para mim, nenhuma é melhor em absoluto, é questão de família sintática.

__Suporte à abstração__. Percebi que as duas têm OO completo, mas Kotlin me impõe o encapsulamento e Python confia na minha disciplina. No exercício, meu private val dados realmente bloqueou o acesso externo em Kotlin; em Python eu usaria _dados ou __dados, que é convenção, não proibição de verdade. Gostei de ter em Kotlin interfaces, classes abstratas, data/sealed class e extension functions ampliando minhas formas de abstrair, enquanto sei que Python me daria ABCs e metaprogramação dinâmica. As duas abstraem bem, mas Kotlin me deu garantias mais rígidas de encapsulamento.

__Expressividade__. Senti um empate, com sotaques diferentes. Em Python eu conto com compreensões de lista e recursos dinâmicos para dizer muito com pouco. Em Kotlin contei com funções de ordem superior, extension functions e sobrecarga de operadores, meu zip { a, b -> a + b } e o 2 * matriz são exemplos do quanto consegui expressar de forma idiomática. A diferença que percebo é que a expressividade de Python é mais dinâmica e a de Kotlin vem acompanhada de segurança de tipos. As duas me deixam ser conciso.


__Checagem de tipos__. Foi talvez a maior diferença que senti. Kotlin é estaticamente e fortemente tipado, com checagem em tempo de compilação e null-safety embutido; aquele meu erro de retornar uma List onde se esperava Matriz2D foi exatamente o que o compilador segurou antes de executar. Python é dinamicamente tipado: a checagem acontece em runtime, e a tipagem estática só existe se eu optar por type hints com ferramentas externas como o mypy, sem o interpretador me obrigar. Para um programa OO maior, concluo que a checagem estática de Kotlin teria evitado muitos bugs que em Python eu só descobriria rodando.

__Manipulação de exceções__. Entendo que Python tem uma cultura de exceções mais central, com o estilo "peça perdão, não permissão" e try/except usado o tempo todo. Em Kotlin eu também tenho try/catch/finally, mas reparei que a ênfase é em pré-condições e null-safety para evitar a exceção; no exercício usei require(...), que lança IllegalArgumentException de forma declarativa, e sei que em Python eu faria raise ValueError ou assert. Vi também que Kotlin oferece o tipo Result para tratar erro sem exceção. Para mim, Python apoia o fluxo nas exceções e Kotlin tenta empurrar a falha para a checagem estática e as pré-condições.

__Restrições de aliases__. Aliasing é quando tenho duas referências apontando para o mesmo objeto mutável, e percebi que Kotlin me dá ferramentas melhores para restringir isso. O val me impede de trocar a referência, a interface List é somente-leitura, e a cópia defensiva map { it.toList() } cortou o vínculo com a lista original, impedindo que um original[0][0] = 999 vazasse para dentro da minha matriz. Entendi que Python sofre o mesmo problema, listas são mutáveis e tudo é referência, mas com menos guardas embutidas: não tenho um val para o conteúdo, e a imutabilidade fica limitada a tuple/frozenset. Por isso meu exercício precisou da cópia defensiva, e levo a lição para as duas linguagens, percebendo que Kotlin me apoia mais para reforçá-la.
